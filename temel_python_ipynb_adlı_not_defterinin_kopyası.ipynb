### 🔸 Soru 5:
`cumle = "  python programlama dili çok güçlüdür  "` değişkeni için:
- Baştaki ve sondaki boşlukları temizleyin (`.strip()`)

### 🔸 Soru 10:
`while` döngüsü ile kullanıcının girdiği sayıya (n=5) kadar Fibonacci dizisini yazdırın.

Örnek çıktı: `0, 1, 1, 2, 3`

```python
# Cevabınızı buraya yazın:

n = 5
x,y = 0, 1 # başlangıç değeri
sayac = 0

while sayac < n:
  print(x, end = " ")
  x, y = y, x + y
  sayac += 1
```

### 🔸 Soru 13 (List Comprehension):
List comprehension kullanarak:
- 1-20 arasındaki çift sayıların listesini oluşturun
- Bir kelime listesindeki 4'ten uzun kelimeleri filtreleyin

```python
# Cevabınızı buraya yazın:

# sonuç - döngü - koşul

# 1-20 arasındaki çift sayılar
cift_sayilar = [ x for x in range(1, 20) if x % 2 == 0]
print(cift_sayilar)


# 4'ten uzun kelimeler
kelimeler = ["at", "kaplan", "it", "fil", "kedi", "aslan", "köpek"]
uzun_kelimeler = [ kelime for kelime in kelimeler if len(kelime) > 4]
print(uzun_kelimeler)
```

### 🔸 Soru 14:
Bir kitap sözlüğü oluşturun (`baslik`, `yazar`, `yil`, `sayfa_sayisi`) ve:
- Tüm anahtar-değer çiftlerini döngüyle yazdırın
- `fiyat` anahtarı ekleyin
- `yazar` anahtarını silin

```python
# Cevabınızı buraya yazın:

# .items(): Sözlükteki hem anahtarı hem de değeri aynı anda döngüye sokmanı sağlar.

# Sözlük
kitap = {
    "baslik": "Notre Dame'ın Kamburu",
    "yazar": "Victor Hugo",
    "yil": 1831,
    "sayfa_sayisi": 550
}

# Döngüyle anahtar-değer çiftleri
for anahtar, deger in kitap.items():
    print(f"{anahtar}: {deger}")

# 'fiyat'
kitap["fiyat"] = 700

# 'yazar'
del kitap["yazar"]

print("\nGüncel Kitap Sözlüğü:", kitap)
```

### 🔸 Soru 15:
Bir sayının asal olup olmadığını kontrol eden `asal_mi(n)` fonksiyonu yazın.
- 1'den 20'ye kadar olan asal sayıları bu fonksiyonu kullanarak yazdırın.

```python
def asal_mi(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# 1'den 20'ye
for i in range(1, 21):
    if asal_mi(i):
        print(i)
```

### 🔸 Soru 16:
Bir listenin ortalamasını, en büyük ve en küçük elemanını döndüren `liste_istatistik(liste)` fonksiyonu yazın.
Fonksiyon 3 değer döndürsün: `(ortalama, minimum, maksimum)`

```python
# ""not liste" ifadesi, "Eğer liste boşsa yani False ise, bunu True yap ve içeri gir
def liste_istatistik(liste):
    if not liste:
        return 0, 0, 0

    ortalama = sum(liste) / len(liste)
    mini = min(liste)
    maksi = max(liste)

    return ortalama, mini, maksi


deneme_liste = [4, 8, 15, 16, 23, 42]
ort, mini, maksi = liste_istatistik(deneme_liste)
print(f"Ortalama: {ort}, Min: {mini}, Max: {maksi}")
```

try-except ile bunu yakalayan ve anlamlı bir mesaj veren kod yazın.

### 🔸 Soru 18 (Bütünleştirici):
Bir **not defteri uygulaması** yazın:
- `notlar` adında boş bir liste oluşturun
- `not_ekle(metin)` fonksiyonu: Notu listeye ekler
- `not_listele()` fonksiyonu: Tüm notları numaralandırarak yazdırır
- `not_sil(index)` fonksiyonu: Verilen indexteki notu siler, hatalı index için hata mesajı verir

Fonksiyonları test edin: 3 not ekleyin, listeleyin, ortadakini silin, tekrar listeleyin.

```python
# Cevabınızı buraya yazın:
notlar = []

def not_ekle(metin):
    notlar.append(metin)
    print(f"Eklendi: {metin}")

def not_listele():
    if not notlar:
        print("Not defteri şu an boş.")
    else:
        print("\nNotlar")
        # enumerate hem notu hem de sırasını (index) bize verir
        for sira, metin in enumerate(notlar, start=1):
            print(f"{sira}. {metin}")

def not_sil(sira_no):
    try:
        # Listeler 0'dan başlar ama kullanıcı 1. notu silmek ister.
        # Bu yüzden sira_no'dan 1 çıkarıyoruz.
        silinen = notlar.pop(sira_no - 1)
        print(f"\nSilindi: {silinen}")
    except IndexError:
        # Eğer listede 3 not varken kullanıcı 5'i silmeye çalışırsa burası çalışır
        print(f"\nHata: {sira_no} numaralı bir not bulunamadı!")

# --- Test Kısmı ---
not_ekle("Python öğren")
not_ekle("Alıştırmaları tamamla")
not_ekle("Proje yap")

not_listele()

not_sil(2) # 2. sıradaki "Alıştırmaları tamamla" notunu siler

not_listele()

# Test kodu:
# not_ekle("Python öğren")
# not_ekle("Alıştırmaları tamamla")
# not_ekle("Proje yap")
# not_listele()
# not_sil(1)
# not_listele()
```
